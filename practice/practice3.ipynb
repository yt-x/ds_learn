# 自动化报告

生成包含结构化表格和数据可视化图片的报告，掌握动态数据插入与格式对齐技巧，确保图文规范统一。  

【任务】  

1.读取Automated_reporting.csv文件中的销售数据（含 “区域”“季度”“销售额” 列），使用pandas加载为df（2分）。  
2.创建 Word 文档，定义规范样式：  
（1）一级标题：宋体、二号字、加粗、段前距 12 磅（2 分）。    
（2）二级标题：宋体、小三号字、加粗、段前距 6 磅（2 分）。    
（3）表格标题：宋体、五号字、居中、段后距 6 磅（2分）。  
3.向文档添加内容：  
（1）一级标题 “2023 季度销售数据报告”（1 分）。  
（2）二级标题 “一、区域季度销售额”，其下插入汇总表格（行：区域，列：季度，值：销售额）（5 分）。  
（3）表格下方添加表格标题 “表 1：各区域季度销售额（单位：万元）”（1 分）。  
4.使用matplotlib生成 “各区域年度总销售额” 柱状图，保存为"output_A/sales_bar.png"（5 分）。  
5.在文档中插入该图片：宽度 14cm、居中对齐，下方添加图片标题 “图 1：区域年度销售额对比”（4 分）。  
6.二级标题 “二、数据总结”，其下添加正文（内容不限，自行分析表格与图片数据编写内容）（5 分）。  
7.保存文档为"docx/sales_report_with_table_image.docx"（1 分）。
