使用正则表达式仅仅保留句子中的中文单词

```python
import re

text = "Hello 世界！Python 很好用。123 中文匹配测试。"
pattern = r'[\u4e00-\u9fff]+'

matches = re.findall(pattern, text)
print(matches)
```

也可以反过来，把非中文字符的全部用空字符替代了

```python
import re

text = "Hello，世界！123 Python 很好用。@#$%^&*（测试）"
# 保留所有在 \u4e00-\u9fff 范围内的汉字，删除其余一切
cleaned = re.sub(r'[^\u4e00-\u9fff]', '', text)
print(cleaned)
```

如果想保留中文标点（如：，。！？；：“”‘’（）【】等），可以扩展 Unicode 范围

```python
import re

text = "Hello，世界！123（测试）"
# 汉字 + 常见中文标点（Unicode 范围）
pattern = r'[^\u4e00-\u9fff\u3000-\u303f\uff00-\uffef]'
cleaned = re.sub(pattern, '', text)
print(cleaned)
# 输出: ，世界！（测试）
```

如果记不住 Unicode 范围（比如 \u4e00-\u9fff），也有其他方法来只保留中文字符

```python
import re

text = "Hello，世界！123 Python 很好用。@#$%^&*（测试）"

# 删除所有英文字母、数字、空白符、标点（包括中英文）
cleaned = re.sub(r'[a-zA-Z0-9\s\W]', '', text)
print(cleaned) # 世界很好用测试
```
